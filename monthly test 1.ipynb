```python
std = {}
```

```python


    
def add():
    id=int(input("ID:"))

    if id in std:
        print("student id already exists")
    else:
        name=input("enter name")
        mark=int(input("mark"))
        std[id]={"name":name,"mark":mark}
        print("student succesfully added")

def view():
    
    print("student list",std)
    
def search():
    searchid=int(input('id:'))
    print(std[searchid])


def update():
    id=int(input("ID:"))
    if id in std:
        newmark=int(input("enter updated mark"))
        std[id]=std["mark"]=newmark
    else:
        print("not found:")
print("1.add student name")
print("2.view all students")
print("3.search students")
print("4.update marks")
print("5. exit")


choice=input("enter choice:")
if choice=="1":
    add()
elif choice=="2":
    view()
elif choice=="3":
    search()
elif choice=="4":
    update()
elif choice=="5":
    print("thankyou! program terminated")
else:
    print("invalid choice")


```
